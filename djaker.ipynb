```python
import heapq
```

```python
graph = {
    "A": [("B", 4), ("C", 2)],
    "B": [("A", 4), ("C", 1), ("D", 5)],
    "C": [("A", 2), ("B", 1), ("D", 5), ("E", 10)],
    "D": [("B", 5), ("C", 5), ("E", 2)],
    "E": [("C", 10), ("D", 2)]
}
```

```python
def dijkstra(grape,start,destination):

    distances = {
        node : float('inf') # A = inf , B = inf , C = inf , D = inf
        for node in grape
    }

    distances[start] = 0 # A = 0 , B = inf , C = inf , D = inf

    previous_node = {

        node : None
        for node in grape
    }

    priority_queue = [] # defined priority queue

    heapq.heappush(
            priority_queue , # pushing the start node to the priority queue
                            # priority_queue = [(0,"A")]

            (0,start)

    )

    while priority_queue: # run the while loop until the priority queue is emply priority_queue = []

        current_distance , current_node = heapq.heappop(
            priority_queue,                                 #  current_distance = 0 , current_node = A

        )

        if current_node == destination: # checking the if statement if the current node is the target node
            break

        for neighbour, weight in graph[current_node]: # for loop for taking the neighbour for the current node , its weight
                                                      # for A neighbourn will be B , C and weight will be 4 , 2

            new_distance = current_distance + weight # createing a new_distance, taking the current distance and adding the 
                                                    # neighbour node weights
                                                    # for B new_distance = 0 + 4
                                                    # for c new_distance = 0 + 2

            if new_distance < distances[neighbour]: # checking if new_distance is smaller then for B = inf if will be trigar
                distances[neighbour] = new_distance # B = inf -> B = 4
                previous_node[neighbour] = current_node # B = None  -> B = A

                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbour)   # pushing the B to priority queue 
                )

            # repeat the same think for every priority queue top elemnt until the pririty quesu of empty
    path = []
    current = destination

    while current is not  None:

        path.append(current)

        current = previous_node[current]

    path.reverse()


    return path, distances[destination] 






```

```python
path, distance = dijkstra(
    graph,
    "E",
    "A"
)

print(path)
print(distance)
```
